```python
import cv2
from google.colab import files

print("Libraries imported successfully")
```

```python
uploaded = files.upload()

filename = list(uploaded.keys())[0]

print("Video uploaded successfully:", filename)
```

```python
cap = cv2.VideoCapture(filename)

frames = []

while True:
    ret, frame = cap.read()

    if not ret:
        break

    frames.append(frame)

cap.release()

print("Total frames:", len(frames))
```

```python
frames.reverse()

print("Video frames reversed successfully")
```

```python
if len(frames) > 0:

    height, width = frames[0].shape[:2]

    fps = 24

    fourcc = cv2.VideoWriter_fourcc(*'mp4v')

    output = cv2.VideoWriter(
        'reversed_video.mp4',
        fourcc,
        fps,
        (width, height)
    )

    for frame in frames:
        output.write(frame)

    output.release()

    print("Reversed video created successfully")

else:
    print("No video frames found")
```
